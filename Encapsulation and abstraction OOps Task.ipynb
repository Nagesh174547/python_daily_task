```python
# food mangement
from abc import ABC, abstractmethod

class Food (ABC):

    #constructor
    def __init__(self,f_name,f_price):
        self.__Food_name = f_name
        self.__Food_price = f_price

    #getter name
    def get_food_name(self):
        return self.__Food_name

   # getter price
    def get_food_price(self):
        return self.__Food_price

    # abstractmethod
    @abstractmethod
    def prepare(self):
        pass

    @abstractmethod
    def deliver(self):
        pass

        
   # display all details
    def display_all_details(self):
        print(f"Food Name: {self.__Food_name}")
        print(f"Food Price: {self.__Food_price}")
        self.prepare()
        self.deliver()
        
        
#child  class
class Pizza(Food):

    def prepare(self):
        print('Pizza is prepared')

    def deliver(self):
        print("Pizza is delivered")


        #child  class
class Biryani(Food):

    def prepare(self):
        print('Biryani is prepared')

    def deliver(self):
        print("Biryani id delivered")


# create the class object

pizza = Pizza("Pizza",500)
biryani = Biryani("Biryani",150)

#pizza.get_food_name()
#biryani.get_food_name()`

pizza.display_all_details()
print("-"*50)
biryani.display_all_details()
```

# 1 Create a class named BankAccount.

```python
from abc import ABC
class BankAcount(ABC):
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self,amt):
        if amt > 0:
            self.__balance += amt
            print("Deposited money :", amt)
        else:
            print("Invalid deposit ammount")

    def withdraw(self, amt):
        if amt > 0 and amt <= self.__balance:
            self.__balance -= amt 
            print("Withdraw money : ", amt)
        else:
            print("Insufficient balance")

    def desplay_balance(self):
        print(f"Current Balance : {self.__balance}")

ac = BankAcount(50000)   # create a object

ac.deposit(30000)

ac.withdraw(70000)

ac.desplay_balance()

```

# 6. Class Name: Product

```python
class Product:
    def __init__(self, price, stock):
        self.__price = price
        self.__stock = stock

    def update_price(self, new_price):
        if new_price >= 0:
            self.__price = new_price
        else:
            print("Price cannot be negative")

    def update_stock(self, qty):
        if self.__stock + qty >= 0:
            self.__stock += qty
        else:
            print("Stock cannot be negative")

# Test
p = Product(100, 10)
p.update_price(120)
p.update_stock(-5)
print(p._Product__price, p._Product__stock)  # 120, 5
```

```python
from datetime import datetime

class Car:
    def __init__(self, model, year):
        self.__model = model
        self.__year = year

    def set_model(self, model):
        self.__model = model

    def get_model(self):
        return self.__model

    def set_year(self, year):
        if year <= datetime.now().year:
            self.__year = year
        else:
            print("Year cannot exceed current year")

    def get_year(self):
        return self.__year

# Test
c = Car("Tesla", 2020)
c.set_year(2026)  # OK if current year ≤ 2026
print(c.get_model(), c.get_year())
```

# 8. Class Name: User

```python
class User:
    def __init__(self, password):
        # store password privately
        self.__password = None
        self.change_password(password)  # validate at creation

    def change_password(self, new_password):
        if len(new_password) >= 8:
            self.__password = new_password
            print("Password updated successfully")
        else:
            print("Password must be at least 8 characters long")

    def get_password(self):
        # just for testing/demo (in real apps you wouldn’t expose password like this!)
        return self.__password


# ✅ Test
u = User("mypassword")          # valid
print(u.get_password())         # "mypassword"

u.change_password("short")      # invalid
print(u.get_password())         # still "mypassword"

u.change_password("newsecurepass")
print(u.get_password())         # "newsecurepass"
```

# 9. Class Name: Rectangle

```python
class Rectangle:
    def __init__(self, length, width):
        self.__length = length
        self.__width = width

    def set_length(self, length):
        if length > 0:
            self.__length = length
        else:
            print("Length must be positive")

    def set_width(self, width):
        if width > 0:
            self.__width = width
        else:
            print("Width must be positive")

    def area(self):
        return self.__length * self.__width

# Test
r = Rectangle(5, 3)
print(r.area())  # 15
```

# 10. Class Name: Account

```python
class Account:
    history = []

    def __init__(self):
        self.__balance = 0

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            Account.history.append(self.__balance)

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            Account.history.append(self.__balance)

# Test
a = Account()
a.deposit(100)
a.withdraw(30)
print(Account.history)  # [100, 70]
```

# 16. Class Name: Shape

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def calculate_area(self): pass

    @abstractmethod
    def calculate_perimeter(self): pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def calculate_area(self):
        return 3.14 * self.radius ** 2

    def calculate_perimeter(self):
        return 2 * 3.14 * self.radius

# Test
c = Circle(5)
print(c.calculate_area(), c.calculate_perimeter())
```

# 25. Class Name: Shape3D 


```python
class Shape3D(ABC):
    @abstractmethod
    def volume(self): pass
    @abstractmethod
    def surface_area(self): pass

class Cube(Shape3D):
    def __init__(self, side): self.side = side
    def volume(self): return self.side ** 3
    def surface_area(self): return 6 * self.side ** 2

class Sphere(Shape3D):
    def __init__(self, r): self.r = r
    def volume(self): return (4/3) * 3.14 * self.r**3
    def surface_area(self): return 4 * 3.14 * self.r**2

# Test
print(Cube(3).volume(), Sphere(2).surface_area())
```

# 27. Class Name: Database 


```python
class Database(ABC):
    @abstractmethod
    def connect(self): pass
    @abstractmethod
    def query(self, sql): pass

class MySQLDatabase(Database):
    def connect(self): print("Connected to MySQL")
    def query(self, sql): print(f"MySQL executing: {sql}")

class MongoDatabase(Database):
    def connect(self): print("Connected to MongoDB")
    def query(self, sql): print(f"Mongo executing: {sql}")

# Test
db = MySQLDatabase()
db.connect()
db.query("SELECT * FROM users")
```
