# create a function calculation (a,b, operation) show how keywod args improve readability

```python
def calculation(a,b,operation):
    if operation == "add":
        return a+b
    elif operation == "subtract":
        return a-b
    elif operation == "multiply":
        return a*b
    elif operation == "divide":
        return a/b

    else:
        return "invalid operation"
calculation(a=10,b=5, operation = "add" )
calculation(10,5, operation = "subtract")
    
```

# Create a function that takes student marks as *args and returns average. 

```python
def student_marks(*args):
    total = sum(args)
    marks = total / len(args)
    return marks
student_marks(55,45)
    
```

# Write a function info(*args) that prints type and value of each argument. 

```python
def info(*nags):
    for value in nags:
        print("Type", type(value), "Value", value)
info("nagesh", 20.3, [1,2,3])
```
