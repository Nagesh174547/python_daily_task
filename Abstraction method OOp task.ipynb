# 1 Vehicle

```python
from abc import ABC,abstractmethod
class vahicle(ABC):
    #abstract method
    @abstractmethod
    def start(self):
        pass
    @abstractmethod
    def stop(self):
        pass

class car(vahicle):                  # child class 1
    def start(self):
        print('Key to start Car')

    def stop(self):                  
        print("Break to stop Car")

class bike(vahicle):                 # child class 2
    def start(self):
        print('Key to start Bike')

    def stop(self):
        print("Break to stop BIke")

# Object class
c = car()
c.start()
c.stop()

b = bike()
b.start()
b.stop()
```

# Super function method 

```python
from abc import ABC,abstractmethod

class Agent(ABC):

    # constructor
    def __init__(self, agent_name, agent_api):
        self.Agent_name = agent_name
        self.__Agent_api = agent_api

    # getter : get api
    def get_agent_api(self):
         return f"Agent Api key : {self.__Agent_api}"


    #abstract method
    @abstractmethod
    def LLM_model(self):
        pass


    #desplay all info
    def agent_info(self):
        return f"Agent Name : {self.Agent_name}\nAgent API : {self.__Agent_api}"



    #child class : 1
class flight_agent(Agent):

        #constructor
    def __init__(self, agent_name, agent_api, role):
        super(). __init__(agent_name, agent_api)
        self.flight_agent_role = role

    # Method
    def LLM_model(self):
        return "open AI LLM model"


    # desplay all methods

    def Display_details(self):
        print(f"Agent role : {self.flight_agent_role}")
        print(self.agent_info())
        print(self.LLM_model())

    # Child class : 2

class Hotel_agent(Agent):

         #constructor
    def __init__(self, agent_name, agent_api, Hotel_agent_role):
        super(). __init__(agent_name, agent_api)
        self.Hotel_agent_role = Hotel_agent_role

    # Method
    def LLM_model(self):
        return "Griba AI LLM model"


    # desplay all methods

    def Display_details(self):
        print(f"Hotel agent role : {self.Hotel_agent_role}")
        print(self.agent_info())
        print(self.LLM_model())


    # create the child class object
Flight =  flight_agent("Nagesh", 80071, "Flight ticket booking")
Hotel  =  Hotel_agent("Ram", 3690, "Hotel book")
    

#print(Flight.get_agent_api())
Hotel.Display_details()
print("*" * 20)   

Flight.Display_details()
    
```

# 2 vahicle

```python
class vahicle(ABC):
    def __init__(self,brand):
        self.Brand = brand

    #abstract method
    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

#child class
class car(vahicle):
    def __init__(self,brand):
        
        super().__init__(brand)

    def start(self):
        return f"car is starting : { self.Brand} "

    def stop(self):
        return f"car is stopping : { self.Brand}"

c = car("toyota")
print(c.start())
print(c.stop())
```

# 3 Animal

```python
class Animal(ABC):
    def __init__(self,name):
        self.Name_dog = name

    #abstract method
    @abstractmethod
    def make_sound(self):
        pass
        

#child class
class dog(Animal):
    def __init__(self,name):
        super().__init__(name)

    def make_sound(self):
        return f"Bhuuuuueeee !: {self.Name_dog} "


d = dog("Tom")
d.make_sound()
```

```python

class Payment(ABC):
    def __init__(self,amount):
        self.Amount = amount

    #abstract method
    @abstractmethod
    def process_payment(self):
        pass

#child class = 1
class CreditCard(Payment):
    def __init__(self,amount):
        super().__init__(amount)

    def process_payment(self):
        return f"CreditCard payment : { self.Amount } "
#child class = 2
class cash(Payment):
    def __init__(self,amount):
        super().__init__(amount)

    def process_payment(self):
        return f"Cash payment: { self.Amount } "

c1 = CreditCard(10000)
print(c1.process_payment())

c2 = cash(20000)
print(c2.process_payment())
```
